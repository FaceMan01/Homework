```python
def caesar_cipher_ru(text, shift, mode="1"):
    # mode: 1 - encrypt, 2 - decrypt

    russian_lower = "абвгдежзийклмнопрстуфхцчшщъыьэюяё"
    russian_upper = russian_lower.upper()
    result = ''
    shift = shift if mode == "1" else -shift
    for char in text:
        if char in russian_lower:
            index = russian_lower.index(char)
            shifted_index = (index + shift) % len(russian_lower)
            result += russian_lower[shifted_index]
        elif char in russian_upper:
            index = russian_upper.index(char)
            shifted_index = (index + shift) % len(russian_upper)
            result += russian_upper[shifted_index]
        else:
            result += char
    return result

def substitution_cipher_ru(text, key, mode="1"):
    # mode: 1 - encrypt, 2 - decrypt

    russian_alphabet_lower = "абвгдежзийклмнопрстуфхцчшщъыьэюяё"
    russian_alphabet_upper = russian_alphabet_lower.upper()
    alphabet = russian_alphabet_lower + russian_alphabet_upper
    if mode == "1":
        mapping = dict(zip(alphabet, key))
    else: # decrypt
        mapping = {v: k for k, v in dict(zip(alphabet, key)).items()} # Инвертируем ключ
    result = ''
    for char in text:
        if char in mapping:
            result += mapping[char]
        else:
            result += char
    return result

def vigenere_cipher_ru(text, key, mode="1"):
    # mode: 1 - encrypt, 2 - decrypt

    russian_lower = "абвгдежзийклмнопрстуфхцчшщъыьэюяё"
    russian_upper = russian_lower.upper()
    key = key.upper()
    key_len = len(key)
    key_index = 0
    result = ''
    for char in text:
        if char in russian_lower:
            shift = russian_upper.index(key[key_index % key_len])
            if mode == "2":
                shift = -shift
            index = russian_lower.index(char)
            shifted_index = (index + shift) % len(russian_lower)
            result += russian_lower[shifted_index]
            key_index += 1
        elif char in russian_upper:
            shift = russian_upper.index(key[key_index % key_len])
            if mode == "2":
                shift = -shift
            index = russian_upper.index(char)
            shifted_index = (index + shift) % len(russian_upper)
            result += russian_upper[shifted_index]
            key_index += 1
        else:
            result += char
    return result

def main():
    while True:
        print("\nВыберите алгоритм:")
        print("1. Шифр Цезаря")
        print("2. Шифр Замены")
        print("3. Шифр Виженера")
        print("0. Выход")

        choice = input("Введите номер алгоритма: ")

        if choice == '0':
            print("Выход из программы.")
            break

        if choice not in ['1', '2', '3']:
            print("Неверный выбор алгоритма. Пожалуйста, выберите 1, 2 или 3.")
            continue

        mode = input("Выберите режим ((1) Шифровка / (2) Дешифровка): ").lower()
        if mode not in ['1', '2']:
            # mode: 1 - encrypt, 2 - decrypt
            print("Неверный режим. Пожалуйста, выберите (1) Шифровка или (2) Дешифровка.")
            continue

        text = input("Введите текст для обработки: ")

        algorithm_name = ""
        key = ""
        shift = 6 # Шаг сдвига для шифра Цезаря

        if choice == '1':
            algorithm_name = "Цезаря"
        elif choice == '2':
            algorithm_name = "Замены"
            key_len = len("абвгдежзийклмнопрстуфхцчшщъыьэюяё") * 2 # для верхнего и нижнего регистра
            key = input(f"Введите ключ (полная перестановка алфавита {key_len} символов, строчные и заглавные): ")
            if len(set(key)) != len(set("абвгдежзийклмнопрстуфхцчшщъыьэюяё" + "АБВГДЕЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯЁ")) or not all(k in "абвгдежзийклмнопрстуфхцчшщъыьэюяёАБВГДЕЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯЁ" for k in key):
                print("Ошибка: Ключ замены должен быть полной перестановкой русского алфавита (строчные и заглавные, 66 символов).")
                continue
        elif choice == '3':
            algorithm_name = "Виженера"
            key = input("Введите ключ (буквенное слово на русском): ")
            if not all(k in "абвгдежзийклмнопрстуфхцчшщъыьэюяёАБВГДЕЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯЁ" for k in key):
                print("Ошибка: Ключ Виженера должен содержать только русские буквы.")
                continue

        result_text = ""
        if choice == '1':
            result_text = caesar_cipher_ru(text, shift, mode)
        elif choice == '2':
            result_text = substitution_cipher_ru(text, key, mode)
        elif choice == '3':
            result_text = vigenere_cipher_ru(text, key, mode)

        if mode == '1':
            print(f"\nЗашифрованный текст ({algorithm_name}): {result_text}")
        else:
            print(f"\nРасшифрованный текст ({algorithm_name}): {result_text}")

if __name__ == "__main__":
    main()
```
